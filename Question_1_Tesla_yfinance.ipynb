# Question 1: Use yfinance to Extract Stock Data

**Task:** Reset the index, save, and display the first five rows of the `tesla_data` dataframe using the `head()` function.

Run the code cell below in order. The final expression displays the first five rows.

```python
# If yfinance is not installed, run this in a separate cell:
# %pip install yfinance

import yfinance as yf

# Create a ticker object for Tesla
tesla = yf.Ticker("TSLA")

# Extract historical stock data
tesla_data = tesla.history(period="max")

# Reset the index
tesla_data.reset_index(inplace=True)

# Save the dataframe as a CSV file
tesla_data.to_csv("tesla_stock_data.csv", index=False)

# Display the first five rows
tesla_data.head()
```
