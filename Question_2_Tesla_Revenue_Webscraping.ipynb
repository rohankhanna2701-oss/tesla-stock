# Question 2: Use Webscraping to Extract Tesla Revenue Data

**Task:** Display the last five rows of the `tesla_revenue` dataframe using the `tail()` function.

This notebook scrapes Tesla revenue information from the Macrotrends page, then displays the final five rows. Website access or page structure can change; if scraping is blocked, use the revenue table supplied by your course.

```python
# If needed, install the libraries in a separate cell:
# %pip install pandas requests lxml html5lib

import pandas as pd
import requests
from io import StringIO

# Tesla revenue history webpage
url = "https://www.macrotrends.net/stocks/charts/TSLA/tesla/revenue"
headers = {"User-Agent": "Mozilla/5.0"}

# Request the webpage
response = requests.get(url, headers=headers, timeout=30)
response.raise_for_status()

# Extract all HTML tables from the webpage
tables = pd.read_html(StringIO(response.text))
print("Number of tables found:", len(tables))

# Find the table that contains Tesla revenue
tesla_revenue = None
for table in tables:
    columns_text = " ".join(str(col).lower() for col in table.columns)
    sample_text = table.astype(str).head(10).to_string().lower()
    if "revenue" in columns_text or "revenue" in sample_text:
        tesla_revenue = table.copy()
        break

if tesla_revenue is None:
    raise ValueError("Could not find a Tesla revenue table on the webpage. Check the page or use the course-provided data.")

# Clean column headings and remove empty rows
tesla_revenue.columns = [str(col).strip() for col in tesla_revenue.columns]
tesla_revenue.dropna(how="all", inplace=True)

# Display the last five rows
tesla_revenue.tail()
```
